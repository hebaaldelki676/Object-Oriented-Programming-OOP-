```python
class Calculator:
  def add(self, a, b):
    return a + b

  def multiply(self, a, b):
    return a * b

  def divide(self, a, b):
    return a / b
  def divide2(self,a,b):
    return a // b

  def subtract(self, a, b):
    return a - b

  def remainder(self, a, b):
    return a % b

calc = Calculator()

print("add="  ,     calc.add(5, 3))
print("multiply" ,  calc.multiply(4, 7))
print("divide" ,    calc.divide(20, 4))
print( "subtract" , calc.subtract(15, 6))
print("divide2" ,   calc.divide2(10.5,2))
print( "remainder", calc.remainder(17, 5))
```
