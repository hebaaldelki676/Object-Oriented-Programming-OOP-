```python
class Point:
    x = 1
    y = 2

    def translate(self, dx, dy):
        self.x += dx
        self.y += dy

    def subtract(self, dx, dy):
        self.x -= dx
        self.y -= dy

    def swap(self, dx, dy):
        self.y = dy
        self.x = dx
p1 = Point()
p1.swap(9, 8)
print(p1.x)
print(p1.y)

p1.subtract(5, 7)
print(p1.x)
print(p1.y)

p1.translate(4, 3)
print(p1.x)
print(p1.y)
```
