```python
class Circle:
    def __init__(self, radius, pi=3.14):
        self.pi = float(pi)
        self.radius = radius

    def area(self):
        return float(self.pi * (self.radius ** 2))

c1 = Circle(radius=5, pi=3.14)
c1.color = 'Red'
print("Pi:", c1.pi)
print("Radius:", c1.radius)
print("Color:", c1.color)
print("Area:", c1.area())
```
