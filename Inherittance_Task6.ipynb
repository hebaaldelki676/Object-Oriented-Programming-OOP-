```python
class Parent:
    def __init__(self, fname, lastname, birthdate):
        self.firstname = fname
        self.lastname = lastname
        self.birthdate = birthdate

    def print(self):
        print("First Name:", self.firstname)
        print("Last Name:", self.lastname)
        print("Birth Date:", self.birthdate)

class Child1(Parent):
    pass

class Child2(Parent):
    def __init__(self, fname, lastname, birthdate):
        super().__init__(fname, lastname, birthdate)

class Child3(Parent):
    def __init__(self, fname, lastname, year):
        super().__init__(fname, lastname, year)

x = Child1("Heba", "Fawwaz", 2007)
y = Child2("Moath", "Fawwaz", 1985)
z = Child3("Moath", "Fawwaz", 2026)

print("Child 1:")
x.print()

print("\nChild 2:")
y.print()

print("\nChild 3:")
z.print()
```
