```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)

class Student(Person):
    pass

class Nurs(Person):
    pass

class Teacher(Person):
    pass
class Doctor(Person):
    pass

y = Student("CHILD2", "INSIDE PRINT")
y.printname()

x = Nurs("Mike","Olsen")
x.printname()

z = Teacher("Ali", "Ahmad")
z.printname()

w = Doctor("Sara", "Mohammad")
w.printname()
```
