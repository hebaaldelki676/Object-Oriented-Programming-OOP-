```python

class Parent:
    def __init__(self, fname, lastname, birthdate):
        self.firstname = fname
        self.lastname = lastname
        self.birthdate = birthdate

    def print(self):
        print(f"Name: {self.firstname} {self.lastname}, Birthdate: {self.birthdate}")


class Child1(Parent):
    pass


class Child2(Parent):
    def __init__(self, fname, lastname, birthdate):
        Parent.__init__(self, fname, lastname, birthdate)


class Child3(Parent):
    def __init__(self, fname, lastname, birthdate, year):
        super().__init__(fname, lastname, birthdate)
        self.year = year

    def print(self):
        print(f"Name: {self.firstname} {self.lastname}, Birthdate: {self.birthdate}, Year: {self.year}")


person1 = Child1("Heba", "Moath", "2007-02-07")
person2 = Child2("Moath", "Fawaz", "1985-12-21")
person3 = Child3("Heba", "Moath Fawaz", "1970_11_01", 2026)

person1.print()
person2.print()
person3.print()
```
